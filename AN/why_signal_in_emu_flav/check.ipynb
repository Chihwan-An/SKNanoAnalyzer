```python
wr20_n1_ee = "/gv0/Users/achihwan/SKNano/Run3NanoAODv12/2023/WRtoNEltoElElJJ_MWR2000_N100_TuneCP5_13p6TeV_madgraph-pythia8/NANOAODSIM/130X_mcRun3_2023_realistic_v15-v2/"
file7 = uproot.open(wr20_n1_ee)
events = file7["Events"]

heepid = events["Electron_cutBased_HEEP"].array()
heepid = (heepid == True )
muon_highpt_id = events["Muon_highPtId"].array()
muon_highpt_id = (muon_highpt_id==2)
heep_pass_el = electron[heepid]
highptid_pass_mu = muon[muon_highpt_id]

el_idpass_num = ak.num(heep_pass_el)
mu_idpass_num = ak.num(highptid_pass_mu)
el_1_filter = (el_idpass_num==1)
mu_1_filter = (mu_idpass_num==1)
both_el_mu_num_filter = el_1_filter&mu_1_filter

heep_pass_el_1 = heep_pass_el[both_el_mu_num_filter]
highpt_pass_mu_1 =highptid_pass_mu[both_el_mu_num_filter]
pt_el_cut = (heep_pass_el_1 > 118)
pt_mu_cut = (highpt_pass_mu_1 > 53)
both_pt_filter = pt_el_cut&pt_mu_cut
both_pt_filter = ak.flatten(both_pt_filter) ## make event 단위로 
#####
#####

muon_pt  = events["Muon_pt"].array() [muon_highpt_id]   # highPtId==2 통과한 것만
muon_eta = events["Muon_eta"].array()[muon_highpt_id]
muon_phi = events["Muon_phi"].array()[muon_highpt_id]

gen_pt = events["GenPart_pt"].array()
gen_eta = events["GenPart_eta"].array()
gen_phi = events["GenPart_phi"].array()

gen_pdgid = events["GenPart_pdgId"].array()
gen_mother_idx = events["GenPart_genPartIdxMother"].array()
gen_statusflag = events["GenPart_statusFlags"].array()

### check filter  ####
#gen_pt = gen_pt[both_el_mu_num_filter]
#print(len(gen_pt))
#print(gen_pt)
#gen_pt = gen_pt[both_pt_filter]
#print(len(gen_pt))
#print(gen_pt)
#######################

### genevents after filter 

muon_pt = muon_pt[both_el_mu_num_filter]
muon_pt = muon_pt[both_pt_filter]

muon_eta = muon_eta[both_el_mu_num_filter]
muon_eta = muon_eta[both_pt_filter]

muon_phi = muon_phi[both_el_mu_num_filter]
muon_phi = muon_phi[both_pt_filter]

gen_pt = gen_pt[both_el_mu_num_filter]
gen_pt = gen_pt[both_pt_filter]

gen_eta = gen_eta[both_el_mu_num_filter]
gen_eta = gen_eta[both_pt_filter]

gen_phi = gen_phi[both_el_mu_num_filter] 
gen_phi = gen_phi[both_pt_filter] 

gen_pdgid = gen_pdgid[both_el_mu_num_filter]
gen_pdgid = gen_pdgid[both_pt_filter]

gen_mother_idx = gen_mother_idx[both_el_mu_num_filter]
gen_mother_idx = gen_mother_idx[both_pt_filter]

gen_statusflag = gen_statusflag[both_el_mu_num_filter]
gen_statusflag = gen_statusflag[both_pt_filter]


####



# ── Gen matching: dR < 0.1 AND |pt_reco - pt_gen| < 10 ──────────────────────

def delta_r(eta1, phi1, eta2, phi2):
    deta = eta1 - eta2
    dphi = phi1 - phi2
    dphi = ak.where(dphi >  np.pi, dphi - 2*np.pi, dphi)
    dphi = ak.where(dphi < -np.pi, dphi + 2*np.pi, dphi)
    return np.sqrt(deta**2 + dphi**2)

# (ev, mu, 1) vs (ev, 1, gen) → (ev, mu, gen)
dR  = delta_r(
    muon_eta[:, :, np.newaxis], muon_phi[:, :, np.newaxis],
    gen_eta [:, np.newaxis, :], gen_phi [:, np.newaxis, :]
)
dPt = np.abs(muon_pt[:, :, np.newaxis] - gen_pt[:, np.newaxis, :])

match_mask = (dR < 0.1) & (dPt < 10.0)   # (ev, mu, gen)

# 조건 불만족 슬롯은 999로 마스킹
dR_masked = ak.where(match_mask, dR, 999.0)

# ── 핵심 수정: ak.argmin → gen axis(axis=2)로 펼쳐서 직접 슬라이싱 ───────────

# best_gen_idx: (ev, mu) — 각 muon에 대해 가장 가까운 gen의 인덱스
best_gen_idx = ak.argmin(dR_masked, axis=2, keepdims=True)  # (ev, mu, 1)
has_match    = ak.min (dR_masked,   axis=2) < 999.0         # (ev, mu)

# gen 배열을 (ev, mu, gen) 로 broadcast한 뒤 best_gen_idx로 슬라이싱
# → ak.Array는 [array, array] 동시 인덱싱 불가이므로
#   axis=2 슬라이싱만 따로 수행
def pick_best_gen(gen_field):
    """gen_field: (ev, gen) → output: (ev, mu) best-matched gen값"""
    broadcast = gen_field[:, np.newaxis, :]          # (ev, 1, gen)
    broadcast = ak.broadcast_arrays(                  # (ev, mu, gen)
        muon_pt[:, :, np.newaxis] * 0,               # muon shape 맞추기용 dummy
        broadcast
    )[1]
    picked = broadcast[best_gen_idx]                  # (ev, mu, 1)
    return ak.firsts(picked, axis=2)                  # (ev, mu)

matched_gen_pt         = pick_best_gen(gen_pt)
matched_gen_eta        = pick_best_gen(gen_eta)
matched_gen_phi        = pick_best_gen(gen_phi)
matched_gen_pdgid      = pick_best_gen(gen_pdgid)
matched_gen_motheridx  = pick_best_gen(gen_mother_idx)
matched_gen_statusflag = pick_best_gen(gen_statusflag)

# 매칭 안 된 muon → -999 처리
matched_gen_pt         = ak.where(has_match, matched_gen_pt,         -999.0)
matched_gen_eta        = ak.where(has_match, matched_gen_eta,        -999.0)
matched_gen_phi        = ak.where(has_match, matched_gen_phi,        -999.0)
matched_gen_pdgid      = ak.where(has_match, matched_gen_pdgid,      -999)
matched_gen_motheridx  = ak.where(has_match, matched_gen_motheridx,  -999)
matched_gen_statusflag = ak.where(has_match, matched_gen_statusflag, -999)

# ── 매칭 성공한 muon만 필터링 ───────────────────────────────────────────────
muon_pt_matched  = muon_pt [has_match]
muon_eta_matched = muon_eta[has_match]
muon_phi_matched = muon_phi[has_match]

gen_pt_matched         = matched_gen_pt        [has_match]
gen_pdgid_matched      = matched_gen_pdgid     [has_match]
gen_motheridx_matched  = matched_gen_motheridx [has_match]
gen_statusflag_matched = matched_gen_statusflag[has_match]

# ── 확인 ────────────────────────────────────────────────────────────────────
n_muon    = ak.sum(ak.num(muon_pt))
n_matched = ak.sum(ak.num(muon_pt_matched))
print(f"Muon total  : {n_muon}")
print(f"Gen matched : {n_matched}  ({100*n_matched/n_muon:.1f}%)")
```

```python
import numpy as np
import awkward as ak

# ── StatusFlags 비트 정의 ────────────────────────────────────────────────────
STATUS_FLAG_BITS = {
    0:  "isPrompt",
    1:  "isDecayedLeptonHadron",
    2:  "isTauDecayProduct",
    3:  "isPromptTauDecayProduct",
    4:  "isDirectTauDecayProduct",
    5:  "isDirectPromptTauDecayProduct",
    6:  "isDirectHadronDecayProduct",
    7:  "isHardProcess",
    8:  "fromHardProcess",
    9:  "isHardProcessTauDecayProduct",
    10: "isDirectHardProcessTauDecayProduct",
    11: "fromHardProcessBeforeFSR",
    12: "isFirstCopy",
    13: "isLastCopy",
    14: "isLastCopyBeforeFSR",
}

# ── 유틸 함수 ────────────────────────────────────────────────────────────────
def to_1d(arr):
    """(ev, 1) or (ev,) → (ev,) 1D로 통일"""
    if arr.ndim == 2:
        return ak.firsts(arr, axis=1)
    return arr

def decode_statusflag(flag_val):
    """정수 statusFlag → 켜진 비트 이름 리스트"""
    if flag_val < 0:
        return ["NO_MATCH"]
    return [name for bit, name in STATUS_FLAG_BITS.items() if (flag_val >> bit) & 1]

def check_bit(flag_array, bit):
    """ak.Array of statusFlags → 특정 비트 bool array"""
    return ((flag_array >> bit) & 1) == 1

def get_mother_field(gen_field, mother_idx):
    """mother_idx (ev,) 로 gen_field에서 mother 값 추출 → (ev,)"""
    mother_idx = to_1d(mother_idx)
    result = []
    for ev_field, ev_midx in zip(gen_field, mother_idx):
        if ev_midx is None or ev_midx < 0:
            result.append(-999)
        else:
            result.append(int(ev_field[ev_midx]))
    return ak.Array(result)

# ── (ev,1) → (ev,) 정규화 ────────────────────────────────────────────────────
muon_pt_matched        = to_1d(muon_pt_matched)
gen_pt_matched         = to_1d(gen_pt_matched)
gen_pdgid_matched      = to_1d(gen_pdgid_matched)
gen_motheridx_matched  = to_1d(gen_motheridx_matched)
gen_statusflag_matched = to_1d(gen_statusflag_matched)

# ── mother 정보 추출 ──────────────────────────────────────────────────────────
mother_pdgid      = get_mother_field(gen_pdgid,      gen_motheridx_matched)
mother_statusflag = get_mother_field(gen_statusflag, gen_motheridx_matched)

# ── 기본 출력 ────────────────────────────────────────────────────────────────
print("=== Basic Info ===")
print(f"  muon_pt_matched        : {muon_pt_matched}")
print(f"  gen_pt_matched         : {gen_pt_matched}")
print(f"  gen_pdgid_matched      : {gen_pdgid_matched}")
print(f"  gen_motheridx_matched  : {gen_motheridx_matched}")
print(f"  gen_statusflag_matched : {gen_statusflag_matched}")

# ── mother pdgid 분포 ────────────────────────────────────────────────────────
print("\n=== Mother pdgid 분포 ===")
flat = ak.to_numpy(mother_pdgid)
unique, counts = np.unique(flat, return_counts=True)
for pid, cnt in sorted(zip(unique, counts), key=lambda x: -x[1]):
    print(f"  pdgid {pid:8d} : {cnt}")

# ── event별 상세 출력 ────────────────────────────────────────────────────────
for ev_i, (mpt, gpt, gpdg, midx, mpdg, sflag, msflag) in enumerate(zip(
    muon_pt_matched,
    gen_pt_matched,
    gen_pdgid_matched,
    gen_motheridx_matched,
    mother_pdgid,
    gen_statusflag_matched,
    mother_statusflag,
)):
    def safe_int(val):
        if val is None:
            return -999
        try:
            return int(val)
        except:
            return -999

    def safe_float(val):
        if val is None:
            return float('nan')
        try:
            return float(val)
        except:
            return float('nan')

    mpt_str  = f"{safe_float(mpt):8.2f}"
    gpt_str  = f"{safe_float(gpt):8.2f}"
    gpdg_i   = safe_int(gpdg)
    midx_str = f"{safe_int(midx):8d}"
    mpdg_str = f"{safe_int(mpdg):8d}"

    mu_flags = decode_statusflag(safe_int(sflag))
    mo_flags = decode_statusflag(safe_int(msflag))

    print(f"  {ev_i:4d} | {mpt_str} | {gpt_str} | {gpdg_i:7d} | "
          f"{midx_str} | {mpdg_str} | {mu_flags} | {mo_flags}")

# ── 비트별 통계 ──────────────────────────────────────────────────────────────
for label, flag_arr in [("Matched Muon", gen_statusflag_matched),
                         ("Mother",       mother_statusflag)]:
    print(f"\n=== {label} — StatusFlag bit 통계 ===")
    valid = flag_arr[flag_arr >= 0]
    total = len(valid)
    for bit, name in STATUS_FLAG_BITS.items():
        n = int(ak.sum(check_bit(valid, bit)))
        print(f"  bit {bit:2d}  {name:40s} : {n:4d} / {total}")
```

```python
import glob, os
import numpy as np
import awkward as ak
import uproot

# ── 설정 ─────────────────────────────────────────────────────────────────────
base_dir   = "/gv0/Users/achihwan/SKNano/Run3NanoAODv12/2023/WRtoNEltoElElJJ_MWR2000_N100_TuneCP5_13p6TeV_madgraph-pythia8/NANOAODSIM/130X_mcRun3_2023_realistic_v15-v2"
root_files = sorted(glob.glob(os.path.join(base_dir, "**/*.root"), recursive=True))
print(f"총 파일 수: {len(root_files)}")

# ── StatusFlag 유틸 ───────────────────────────────────────────────────────────
STATUS_FLAG_BITS = {
    0:"isPrompt", 1:"isDecayedLeptonHadron", 2:"isTauDecayProduct",
    3:"isPromptTauDecayProduct", 4:"isDirectTauDecayProduct",
    5:"isDirectPromptTauDecayProduct", 6:"isDirectHadronDecayProduct",
    7:"isHardProcess", 8:"fromHardProcess", 9:"isHardProcessTauDecayProduct",
    10:"isDirectHardProcessTauDecayProduct", 11:"fromHardProcessBeforeFSR",
    12:"isFirstCopy", 13:"isLastCopy", 14:"isLastCopyBeforeFSR",
}

def to_1d(arr):
    return ak.firsts(arr, axis=1) if arr.ndim == 2 else arr

def check_bit(flag_array, bit):
    return ((flag_array >> bit) & 1) == 1

def decode_statusflag(flag_val):
    if flag_val < 0:
        return ["NO_MATCH"]
    return [name for bit, name in STATUS_FLAG_BITS.items() if (flag_val >> bit) & 1]

def safe_int(val):
    if val is None: return -999
    try: return int(val)
    except: return -999

def safe_float(val):
    if val is None: return float('nan')
    try: return float(val)
    except: return float('nan')

def delta_r(eta1, phi1, eta2, phi2):
    deta = eta1 - eta2
    dphi = phi1 - phi2
    dphi = ak.where(dphi >  np.pi, dphi - 2*np.pi, dphi)
    dphi = ak.where(dphi < -np.pi, dphi + 2*np.pi, dphi)
    return np.sqrt(deta**2 + dphi**2)

def get_mother_field(gen_field, mother_idx):
    mother_idx = to_1d(mother_idx)
    result = []
    for ev_field, ev_midx in zip(gen_field, mother_idx):
        if ev_midx is None or ev_midx < 0:
            result.append(-999)
        else:
            result.append(int(ev_field[ev_midx]))
    return ak.Array(result)

# ── 결과 누적용 리스트 ────────────────────────────────────────────────────────
all_muon_pt          = []
all_gen_pt           = []
all_gen_pdgid        = []
all_gen_motheridx    = []
all_gen_statusflag   = []
all_mother_pdgid     = []
all_mother_statusflag= []

total_muon   = 0
total_matched= 0

# ── 파일 루프 ─────────────────────────────────────────────────────────────────
for i_file, fpath in enumerate(root_files):
    print(f"\n[{i_file+1}/{len(root_files)}] {os.path.basename(fpath)}")

    try:
        file_  = uproot.open(fpath)
        events = file_["Events"]
    except Exception as e:
        print(f"  !! 파일 열기 실패: {e}")
        continue

    # ── ID 필터 ──────────────────────────────────────────────────────────────
    electron      = events["Electron_pt"].array()
    muon          = events["Muon_pt"].array()
    heepid        = events["Electron_cutBased_HEEP"].array() == True
    muon_highpt_id= events["Muon_highPtId"].array() == 2

    heep_pass_el    = electron[heepid]
    highptid_pass_mu= muon[muon_highpt_id]

    el_1_filter  = (ak.num(heep_pass_el)     == 1)
    mu_1_filter  = (ak.num(highptid_pass_mu) == 1)
    both_el_mu_num_filter = el_1_filter & mu_1_filter

    heep_pass_el_1   = heep_pass_el    [both_el_mu_num_filter]
    highpt_pass_mu_1 = highptid_pass_mu[both_el_mu_num_filter]

    pt_el_cut = heep_pass_el_1   > 118
    pt_mu_cut = highpt_pass_mu_1 > 53
    both_pt_filter = ak.flatten(pt_el_cut & pt_mu_cut)

    # ── 배열 로드 및 필터 적용 ────────────────────────────────────────────────
    def load_and_filter(branch, id_mask=None):
        arr = events[branch].array()
        if id_mask is not None:
            arr = arr[id_mask]
        arr = arr[both_el_mu_num_filter]
        arr = arr[both_pt_filter]
        return arr

    muon_pt  = load_and_filter("Muon_pt",  muon_highpt_id)
    muon_eta = load_and_filter("Muon_eta", muon_highpt_id)
    muon_phi = load_and_filter("Muon_phi", muon_highpt_id)

    gen_pt        = load_and_filter("GenPart_pt")
    gen_eta       = load_and_filter("GenPart_eta")
    gen_phi       = load_and_filter("GenPart_phi")
    gen_pdgid     = load_and_filter("GenPart_pdgId")
    gen_mother_idx= load_and_filter("GenPart_genPartIdxMother")
    gen_statusflag= load_and_filter("GenPart_statusFlags")

    if ak.sum(ak.num(muon_pt)) == 0:
        print("  통과 이벤트 없음, skip")
        continue

    # ── Gen matching ─────────────────────────────────────────────────────────
    dR  = delta_r(
        muon_eta[:, :, np.newaxis], muon_phi[:, :, np.newaxis],
        gen_eta [:, np.newaxis, :], gen_phi [:, np.newaxis, :]
    )
    dPt = np.abs(muon_pt[:, :, np.newaxis] - gen_pt[:, np.newaxis, :])

    dR_masked    = ak.where((dR < 0.1) & (dPt < 10.0), dR, 999.0)
    best_gen_idx = ak.argmin(dR_masked, axis=2, keepdims=True)
    has_match    = ak.min(dR_masked, axis=2) < 999.0

    def pick_best_gen(gen_field):
        broadcast = ak.broadcast_arrays(
            muon_pt[:, :, np.newaxis] * 0,
            gen_field[:, np.newaxis, :]
        )[1]
        return ak.firsts(broadcast[best_gen_idx], axis=2)

    matched_gen_pt        = ak.where(has_match, pick_best_gen(gen_pt),         -999.0)
    matched_gen_pdgid     = ak.where(has_match, pick_best_gen(gen_pdgid),      -999)
    matched_gen_motheridx = ak.where(has_match, pick_best_gen(gen_mother_idx), -999)
    matched_gen_statusflag= ak.where(has_match, pick_best_gen(gen_statusflag), -999)

    muon_pt_m   = to_1d(muon_pt[has_match])
    gen_pt_m    = to_1d(matched_gen_pt[has_match])
    gen_pdg_m   = to_1d(matched_gen_pdgid[has_match])
    gen_midx_m  = to_1d(matched_gen_motheridx[has_match])
    gen_sflag_m = to_1d(matched_gen_statusflag[has_match])

    mo_pdgid  = get_mother_field(gen_pdgid,      gen_midx_m)
    mo_sflag  = get_mother_field(gen_statusflag,  gen_midx_m)

    # 누적
    all_muon_pt          .append(muon_pt_m)
    all_gen_pt           .append(gen_pt_m)
    all_gen_pdgid        .append(gen_pdg_m)
    all_gen_motheridx    .append(gen_midx_m)
    all_gen_statusflag   .append(gen_sflag_m)
    all_mother_pdgid     .append(mo_pdgid)
    all_mother_statusflag.append(mo_sflag)

    n_mu  = int(ak.sum(ak.num(muon_pt)))
    n_mat = int(len(muon_pt_m))
    total_muon    += n_mu
    total_matched += n_mat
    print(f"  muon: {n_mu}  matched: {n_mat}  ({100*n_mat/n_mu:.1f}%)" if n_mu else "  muon 0")

# ── 전체 결과 합치기 ──────────────────────────────────────────────────────────
muon_pt_all          = ak.concatenate(all_muon_pt)
gen_pt_all           = ak.concatenate(all_gen_pt)
gen_pdgid_all        = ak.concatenate(all_gen_pdgid)
gen_motheridx_all    = ak.concatenate(all_gen_motheridx)
gen_statusflag_all   = ak.concatenate(all_gen_statusflag)
mother_pdgid_all     = ak.concatenate(all_mother_pdgid)
mother_statusflag_all= ak.concatenate(all_mother_statusflag)

print(f"\n{'='*60}")
print(f"전체 muon    : {total_muon}")
print(f"전체 matched : {total_matched}  ({100*total_matched/total_muon:.1f}%)")

# ── 최종 통계 ─────────────────────────────────────────────────────────────────
print("\n=== Mother pdgid 분포 (전체) ===")
flat = ak.to_numpy(mother_pdgid_all)
unique, counts = np.unique(flat, return_counts=True)
for pid, cnt in sorted(zip(unique, counts), key=lambda x: -x[1]):
    print(f"  pdgid {pid:8d} : {cnt}")

for label, flag_arr in [("Matched Muon", gen_statusflag_all),
                         ("Mother",       mother_statusflag_all)]:
    print(f"\n=== {label} — StatusFlag bit 통계 (전체) ===")
    valid = flag_arr[flag_arr >= 0]
    total = len(valid)
    for bit, name in STATUS_FLAG_BITS.items():
        n = int(ak.sum(check_bit(valid, bit)))
        print(f"  bit {bit:2d}  {name:40s} : {n:4d} / {total}")
# ── 전체 합치기 이후 아래 추가 ────────────────────────────────────────────────

print("\n=== Event-by-Event Detail (전체) ===")
print(f"  {'ev':>5} | {'mu_pt':>8} | {'gen_pt':>8} | {'pdgid':>7} | "
      f"{'moth_idx':>8} | {'moth_pdg':>8} | muon_flags | mother_flags")
print("  " + "-"*110)

for ev_i, (mpt, gpt, gpdg, midx, mpdg, sflag, msflag) in enumerate(zip(
    muon_pt_all,
    gen_pt_all,
    gen_pdgid_all,
    gen_motheridx_all,
    mother_pdgid_all,
    gen_statusflag_all,
    mother_statusflag_all,
)):
    mpt_str  = f"{safe_float(mpt):8.2f}"
    gpt_str  = f"{safe_float(gpt):8.2f}"
    gpdg_i   = safe_int(gpdg)
    midx_str = f"{safe_int(midx):8d}"
    mpdg_str = f"{safe_int(mpdg):8d}"
    mu_flags = decode_statusflag(safe_int(sflag))
    mo_flags = decode_statusflag(safe_int(msflag))

    print(f"  {ev_i:5d} | {mpt_str} | {gpt_str} | {gpdg_i:7d} | "
          f"{midx_str} | {mpdg_str} | {mu_flags} | {mo_flags}")
```
