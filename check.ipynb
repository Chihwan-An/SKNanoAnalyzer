```python
import glob

sample_configs = [
    {
        "pattern": "/gv0/Users/achihwan/SKNano/Run3NanoAODv13p1/2018/4000100ee/*1.root",
        "label": "signal sample"
    },
    {
        "pattern": "/gv0/DATA/SKNano/NanoAODv15/2024/TTLL_Bin-MLL-50_TuneCP5_13p6TeV_amcatnlo-pythia8/crab_MC_2024_TTLL_Bin-MLL-50_TuneCP5_13p6TeV_amcatnlo-pythia8/251121_124032/0000/*1.root",
        "label": "tt ll sample"
    },
    {
        "pattern": "/gv0/DATA/SKNano/Run3NanoAODv13p1/2023/Muon0/Run2023C_v1/250622_221441/0000/*1.root",
        "label": "Muon0 sample"
    },
]

def process_sample(filepath, label):
    file = uproot.open(filepath)
    events = file["Events"]

    reltun = events["Muon_tunepRelPt"].array()
    muon_pt = events["Muon_pt"].array()

    over200 = muon_pt > 200
    muon_pt = muon_pt[over200]
    reltun = reltun[over200]
    reltun = ak.flatten(reltun, axis=1)

    notone = reltun != 1.0
    reltun = reltun[notone]

    aa = np.unique(reltun)
    print(f"[{label}] {os.path.basename(filepath)}")
    print(f"  exists: {aa}")
    print(f"  muon_pt: {ak.values_astype(muon_pt, np.float64)}")
    print(f"  reltun:  {ak.values_astype(reltun, np.float64)}")
    print("=========================================")


for config in sample_configs:
    files = sorted(glob.glob(config["pattern"]))
    print(f"\n>>> {config['label']}: {len(files)} files found")
    for f in tqdm(files, desc=config["label"]):
        process_sample(f, config["label"])
```
